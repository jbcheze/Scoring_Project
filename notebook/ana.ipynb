# 🧹 Étapes de Nettoyage et d’Inspection des Données

## 🧭 1. Inspection initiale
**Objectif :** comprendre la structure générale de la base.

- Charger les données (`pd.read_csv`, `pd.read_excel`, etc.)
- Examiner les dimensions : `df.shape`
- Aperçu rapide : `df.head()`, `df.info()`, `df.describe()`
- Vérifier les types (`dtypes`) et la cohérence (ex. dates bien en `datetime`, numériques bien en `float`/`int`)

---

## 🔍 2. Analyse des valeurs manquantes
**Objectif :** identifier les colonnes ou lignes à problème.

- Comptage : `df.isna().sum()`
- Proportion manquante : `df.isna().mean()`
- Visualisation possible : `sns.heatmap(df.isna(), cbar=False)`

**Décision :**
- Supprimer si trop de valeurs manquantes (> 50 %)
- Imputer (moyenne, médiane, mode, `ffill`/`bfill`, `KNNImputer`…)

---

## ⚖️ 3. Détection et traitement des doublons
**Objectif :** éviter le biais et les répétitions.

- `df.duplicated().sum()`
- `df = df.drop_duplicates()` si nécessaire.

---

## 📏 4. Recherche de valeurs extrêmes (outliers)
**Objectif :** repérer les observations aberrantes.

**Méthodes :**
- Statistiques : IQR, Z-score (`scipy.stats.zscore`)
- Visualisation : boxplot, scatterplot, histogrammes  

**Décision :**
- Supprimer, corriger ou “winsoriser” (remplacer par les bornes)

---

## 🧮 5. Analyse et nettoyage des incohérences
**Objectif :** assurer la cohérence logique des données.

- Vérifier les unités (ex. salaire en euros ou k€ ?)
- Corriger les fautes de frappe ou les majuscules (`str.lower()`, `strip()`)
- Harmoniser les catégories (`value_counts()`, `replace()`)

---

## 🧠 6. Traitement des types et normalisation
**Objectif :** préparer les variables pour le modèle.

**Conversion :**
- Dates → `datetime`
- Catégorielles → `category` ou encodage (`OneHotEncoder`, `LabelEncoder`)
- Normalisation/standardisation des variables continues si besoin

---

## 🧩 7. Vérification finale
**Objectif :** s’assurer que la base est prête pour l’analyse ou le modèle.

- Plus de valeurs manquantes ?
- Types corrects ?
- Distributions cohérentes ?
- Variables explicatives et cible prêtes ?
